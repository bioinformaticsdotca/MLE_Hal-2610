```python
###imports
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import plot_tree
from sklearn.metrics import classification_report
from sklearn.metrics import ConfusionMatrixDisplay
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, roc_curve, roc_auc_score
from scipy.stats import shapiro
import matplotlib.pyplot as plt
import joblib
```

```python
###lets load the cleveland heart disease data set
df = pd.read_csv(
    "data/processed.cleveland.data",
    header = None)

###naming columns for better visualization
columns = [
    "age",
    "sex",
    "chest_pain_type",
    "resting_blood_pressure",
    "cholesterol",
    "fasting_blood_sugar",
    "resting_ecg",
    "max_heart_rate",
    "exercise_angina",
    "st_depression",
    "st_slope",
    "num_major_vessels",
    "thalassemia",
    "heart_disease"
]
df.columns = columns

###create a binary outcome
df['binary_outcome'] = (df['heart_disease'] > 0).astype(int)

###treat missing data on the X
df = df.replace("?", np.nan)

X = df.drop(columns=['heart_disease', 'binary_outcome'])
y = df['binary_outcome']

df = df.apply(pd.to_numeric)

X = df.drop(columns=['heart_disease', 'binary_outcome'])
y = df['binary_outcome']

###train and test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size = 0.20,
    random_state=42,
    stratify = y
)

imputer = SimpleImputer(strategy="median")


X_train_imputted = pd.DataFrame(
    imputer.fit_transform(X_train),
    columns = X_train.columns,
    index=X_train.index
)

X_test_imputted = pd.DataFrame(
    imputer.transform(X_test),
    columns = X_test.columns,
    index=X_test.index
)


###instantiate RF model (with the best parameters identified in Module 2)
###RF Best parameters:  {'max_depth': 5, 'max_features': 'sqrt', 'min_samples_leaf': 1, 'n_estimators': 50}
rf = RandomForestClassifier(
    max_depth=5,
    max_features='sqrt',
    min_samples_leaf=1,
    n_estimators=50,
    random_state=42)
###fit RF model
rf.fit(X_train_imputted, y_train)

```

```python
###save model
joblib.dump(rf, "rf_model.pkl")
```

```python
###load model
model = joblib.load("rf_model.pkl")

###use loaded model
y_pred = model.predict(X_test_imputted)
print(y_pred)
```
